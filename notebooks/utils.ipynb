# ln(x) and -ln(x)

Plot of the natural logarithm and its negation for positive x.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0.01, 5, 500)
y_ln = np.log(x)
y_neg_ln = -np.log(x)

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(x, y_ln, label=r"$\ln x$", linewidth=2)
ax.plot(x, y_neg_ln, label=r"$-\ln x$", linewidth=2)
ax.axhline(0, color="black", linewidth=0.8, linestyle="--", alpha=0.5)
ax.axvline(1, color="black", linewidth=0.8, linestyle="--", alpha=0.5)
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title(r"$\ln x$ and $-\ln x$")
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```
